```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader  

# Tạo dữ liệu giả
# 100 ảnh, mỗi ảnh có shape 1x28x28
feature_volume_tensor_x = torch.randn(100,1,28,28)

#Tạo nhãn giả
# mỗi nhãn là value tu 0 đến 9

lable_y = torch.randint(0,10,(100,))
print(lable_y)

# Gop x va y lại thành 1 dataset
dataset = TensorDataset(feature_volume_tensor_x,lable_y)

# Tao dataLoader 
# Batch_size = 16 => nghĩa là mỗi lần lấy ra 16 ảnh để train
# shuffle = True => Nghĩa là trộn dữ liệu mỗi epoch
dataloader = DataLoader(dataset=dataset, batch_size= 16, shuffle=True)


for batch_x, batch_y in dataloader:
    # In shape của một bức ảnh
    print("Shape cua ảnh", batch_x.shape)

    # In shape của nhẵn
    print("Shape của nhãn",batch_y.shape)
    break
```

```python
import torch 
from torch.utils.data import DataLoader, TensorDataset

# Tạo dữ liệu giả gồm 95 ảnh xám
feature_volume_tensor = torch.randn(95,1,28,28)

# Giả lặp có 95 nhãn từ 0 đến 9
lable_y = torch.randint(0,10,(95,))
print(lable_y)

# Gộp ảnh vàn nhãn thành dataset
dataset_95 = TensorDataset(feature_volume_tensor, lable_y)

#Chia dữ liệu thành từng batch 20 ảnh
dataloader_20 = DataLoader(dataset=dataset_95, batch_size=20, shuffle=True)

# len(dataLoader) cho biết có bao nhiêu batch trong một epoch

print("Số lượng epoch", len(dataloader_20))

# Duyệt qua từng batch
for i, (batch_x, batch_y) in enumerate(dataloader_20):
    print(f"Số thứ tự batch {i}")
    print(batch_x.shape)
    print(batch_y.shape)
```
