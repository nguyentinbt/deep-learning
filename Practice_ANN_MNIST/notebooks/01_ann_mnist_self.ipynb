```python
import torch
import torch.nn as nn

#Taoj model CNN 
class SimpleCNN(nn.Module):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        # Anh dau vao MNIST co 1 channel do la anh xam
        # Output sau Conv1 se la anh co 8 channel
        # Padding = 1 de giu nguyen kich thươic
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=8, kernel_size=3, padding=1)

        # MãPooling = 2x2 se giam chieu anh di 2 lần
        self.pool = nn.MaxPool2d(2)

        # Nhan 8 channel tu lop conv1 tao ra 16 channel o lop cpnv2
        self.conv2 = nn.Conv2d(in_channels=8, out_channels=16, kernel_size=3, padding=1)

        # Sau 2 lan pooling thi 28x28 => 14x14 => 7x7
        # Shape dau ra luc nay la (N, 16, 7, 7)
        # Flatten se thanh 16 x 7 x 7 = 784
        self.flatten_layer = nn.Flatten()
        self.fc = nn.Linear(784,10)


    def forward(self,tensor_dau_vao):
        tensor_dau_vao = self.conv1(tensor_dau_vao)
        tensor_dau_vao = self.pool(tensor_dau_vao)
        
```
