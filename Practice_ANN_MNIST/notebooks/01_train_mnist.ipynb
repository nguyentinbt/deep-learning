```python
# Cài đặt đường dẫn để import từ thư mục src
import sys
sys.path.append('../')

import torch
import torch.optim as optim
import torch.nn as nn
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Import các module đã tách
from models.model import SimpleANN
from src.train import train_model, evaluate_model

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1. Tải và chuẩn bị dữ liệu (Lưu thẳng vào thư mục data/MNIST/raw)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

train_dataset = datasets.MNIST(root='../data', train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root='../data', train=False, download=True, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=128, shuffle=False)

# 2. Khởi tạo mô hình & Loss & Optimizer
model = SimpleANN().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 3. Chạy Training Loop
train_model(model, train_loader, criterion, optimizer, device, epochs=5)
evaluate_model(model, test_loader, device)

# 4. Lưu mô hình vào thư mục models
torch.save(model.state_dict(), "../models/mnist_ann_best.pth")
```
