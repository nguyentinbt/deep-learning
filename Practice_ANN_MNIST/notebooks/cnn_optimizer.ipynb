```python
import torch 
import torch.nn as nn
import torch.optim as optim


# Khai bao model
# Flow cho training mode: image batch =--> Model --> Loss --> backward --> optimizer 

# Tao model sample Linear co 4 input ra 10 output
model = nn.Linear(in_features=4, out_features=10)

# Tinh loss. Do model nay se du doan so va kq tra ve la 1 mang 10 xac suat , toi se phai can dung mot loss nao do co softmax => CrossEntropy
criterion = nn.CrossEntropyLoss()

# Da co loss gio can optimze
optimizer = optim.Adam(model.parameters(),lr=0.001)


# Su dung
# Tao du lieu gia co 3 mau, moi mau co 4 dac trung
x = torch.randn(3,4)
print(x)

# Tao ra nhan dung cho 3 mau
y = torch.tensor([0,1,0])
print(y)

# Xoa Gradient cu truoc moi lan train
optimizer.zero_grad()

# Du doan dau ra tu model 
output = model(x)

# Tin loss giua du doan va nhan dung
loss = criterion(output,y)

# Tinh Gradient
loss.backward()

# Cap nhat trong so 
optimizer.step()

# in loss
print(loss.item())
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Cac buo de train model
# Dinh nghia mo hinh va cac yeu to cho training: 
# Cac buoc chay training mohinh: Batch --> Model --> Loss -> Backward --> Optimizer
# Du lieu dau vao: bacth anh: tensor 



# Mo hinh : Linear hoac Rergession
# Linear de tinh toan tuyen tinh
model = nn.Linear(in_features=4, out_features=3)
# Loss: Do sai giua gia trij that va du doan
criterion = nn.CrossEntropyLoss()

# Backward de tinh Gradient 
# Optimizer de cap nhat trong so moi dac trung nham giam Loss
optimizer = optim.Adam(model.parameters(),lr=0.001)


# Tao du lieu gia lap
# VD bacth cau cao voi 3 anh 4x4
x = torch.randn(3, 4)

# Label chua gia tri tha cho 3 tam anh
y = torch.randint(0,2,(3,))
print(y)

# Chay model 

# B1: Clear tat ca gradient cu
criterion.zero_grad()

#B1: Dua hinh anh vao model
y_predict = model(x)
print("Predicted Values: ",y_predict)


# Tinh Loss(predict vs actual)
loss = criterion(y_predict, y)
print("Loss value: ",loss)

loss.backward()

# in grad va shape:
print(model.weight.grad.shape)
print(model.weight.shape)
```

```python
### On lai
# Để tạo một model ta cần thúc hiwjwn các bước sau
# 1 Chuẩn bị Batch data -> đây la các bact dang tensor
# 2 Chuẩn bị model: Linear 
# 3 chuẩn bị hàm loss
# tính Gradient bằng backward 
# optimize mô hình bằng Adam hoặc SGD

######
# Import thư thiện
import torch
import torch.nn as nn
import torch.optim as optim

# Tạo sample data. Ví dụ ta có 1 batch 10 ảnh 28 x 28
x_features = torch.randn(10, 28, 28)
y_labled = torch.randint(0,2,(10,))
print(y_labled)
print(y_labled.shape)


# chuẩn bị mô hình. Owr ddaay demo voiws Linear đồ vô là tensor đã flattent
model = nn.Linear(in_features=784,out_features=2)

# Chuẩn bị hàm loss. với hàm loss ta cần cân nhắc dữ liệu đầu ra có bao nhiêu nhãn và mục đích mỗi nhãn như thế nào. ở đây ta có 2 nhãn phân biết True/False và demo với CrossEntropyloss
criterion = nn.CrossEntropyLoss()

# Chuẩn bị cho hàm ompimize bằng cách định nghĩa optimizer với Adam kem learning rate. lr=0.001 là golder rate cho Adam
optimizer = optim.Adam(params=model.parameters(), lr=0.001)

#######
# Chạy mô hình
# Để chạy training mô hình ta có các bước: Bacth --> Model --> Loss --> Backward --> Optimize

# Xóa giá trị cũ trong .grad nếu có để đảm bảo mỗi lần chạy giá trị trong .grad đều là mới
optimizer.zero_grad()

# băt đầu đưa batch vào mô hình (chạy mộ hình) - Forward
# Trước tiên phải kéo phẳng bâtch 10 ảnh đầu vào 
# Khai báo kiêu flattent bằn class Flatten hay hàm view()
flatten = nn.Flatten()
x_flatten =flatten(x_features)
y_predict = model(x_flatten)
print(y_predict)

# Sau khi đã cho batch ảnh qua model, giờ tinhd loss
loss = criterion(y_predict, y_labled)
print(loss)

# Đã có loss thì bắt đầu tính đạo hàm của loss bằng cách gọi hàm backward. Trong backward này, nó trả về trọng số mới của Gradient desent và chỉ lưu ở .grad 
loss.backward()

print("Shape cua model",model.weight.shape)
print("Shape cua Gradient descent", model.weight.grad.shape)

# Sau khi tính xong Gradien, ta bắt đầu optimize hay còn gọi là áp dụng các trọng số mơi mà backward đã tìm được và mô hình. Backpropagration
optimizer.step()

print("Model dau khi optimize", model.weight.shape)

```
