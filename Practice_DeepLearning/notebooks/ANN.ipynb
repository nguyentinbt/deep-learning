```python
import torch
import torch.nn as nn

device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")

# Định nghĩa lớp Linear và ReLu
model = nn.Linear(in_features=4, out_features=2).to(device)
relu = nn.ReLU()

# Khai báo dữ liệu tensor ngẫu nhiên
x_feature = torch.randn(2, 4)

flatten = nn.Flatten()
x_flatten = flatten(x_feature)
# Gọi model và truyền dữ liệu
out_linear = model(x_flatten)
out_relu = relu(out_linear)
```
